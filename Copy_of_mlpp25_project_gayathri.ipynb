U.S.  Then, the team follows these steps:
1. Assign weights: First, counties are ranked within each of the 50 U.S. states using measures that are assigned
relative weights within the Health Outcomes and Health Factors Areas.
2. Calculate Z-scores: Next, measures are standardized within each state using a Z-score.  
3. Create composite scores: The ranks are then calculated based on weighted sums of the standardized measures
within each state—we call these weighted sums a composite score. The county with the lowest composite
score (best health) gets a rank of #1 for that state and the county with the highest composite score (worst
health) is assigned the lowest rank.  
4. Calculate Ranks and quartiles: After the Ranks are calculated, CHR&R calculates quartiles. Counties are
grouped into four equally sized groups (quartiles) from Least Healthy to Healthiest (Lowest 0-25%, Lower 25
50%, Higher 50-75%, and Healthiest 75-100%) within each state.  
5. Create supplemental tools: Lastly, we calculate trend and Areas to Explore and Areas of Strength.

To replace NaN missing values, I tried two imputation methods: mean and median.

Mean imputation uses the average of all observed values. Because Non-Hispanic_Black_pct is right-skewed with outliers, its mean is 0.09, which pulls the imputed values toward extremes.

Median imputation sorts the data and picks the middle value, so it isn't influenced by outliers. For Non-Hispanic_Black_pct, the median is 0.02.

The difference between mean and median is 0.07, showing how mean imputation can inject extreme values. Therefore, we proceed with median imputation.

```python
# also checkk the mean, and if both are different then outliers are present, so replace the outliers with median

# -- replalcing nan values with the median
num_cols_analytics = analytics_23_df.select_dtypes(include=['float64']).columns
col_median = analytics_23_df[num_cols_analytics].median()
analytics_23_df.loc[:, num_cols_analytics] = analytics_23_df[num_cols_analytics].fillna(col_median)

print(analytics_23_df.isna().sum())
```

#### Data Preparation-Regression Model

First split the data into training and test sets, then on the training data only we handle extreme values (beyond 1.5 times IQR) so a few outliers don't skew our model. Next we standardize every numeric feature to zero mean and unit variance so that all inputs weigh equally and optimization runs smoothly. Finally, we apply those same outlier caps and scaling rules to the test set and use it held completely separate to evaluate real performance and avoid overfitting.

By follwoing these steps we avoid the data leakage problem.

```python
# -- creating list of input features

feat_cols = [
       'Physical_Inactivity_pct',
       'Median_Household_Income',
       'Below_age_18_pct',
       'age_above_eq_65_pct',
       'Non-Hispanic_Black_pct',
       'American_Indian_or_Alaska Native_pct',
       'Asian_pct',
       'race_other_pct',
       'Hispanic_pct',
       'Non-Hispanic_White_pct',
       'Adult_Smoking_pct',
       'Adult_Obesity_pct',
       'Food_Environment_Index',
       'Access_to_Exercise_Opportunities_pct',
       'Excessive_Drinking_pct',
       'Severe_Housing_Problems_pct',
       'Air_Pollution_Particulate_Matter',
       'Diabetes_Prevalence_pct',
       'Food_Insecurity_pct',
       'Uninsured_pct',
       'Children_in_Single-Parent_Households_pct',
       'Insufficient_Sleep_pct',
       'Mental_Health_Providers_ratio',
       'Educ_affir_pct',
       'Pop_est_val',
       'Proverty_pct',
       'unemployement_pct',
       'Food_Access_Difficulty',
       'Demographic_Vulnerability_Ratio'
      ]

# -- create input features
feat = combined_23_data_df[feat_cols]
feat.columns

# -- create target features
targ = combined_23_data_df["Poor_Mental_Health_Days"]
```

```python
# -- compute pearson correlation co-efficient between the input and target features
for col in feat_cols:
    df_pair = feat[[col]].join(targ).dropna()
    xi = df_pair[col].values
    yi = df_pair["Poor_Mental_Health_Days"].values

    cov = np.cov(xi, yi, bias=True)[0, 1]
    corr = cov / (np.std(xi) * np.std(yi))

    plt.figure(figsize=(5, 4))
    plt.scatter(xi, yi, s=10)
    plt.title(f"{col} vs Poor_Mental_Health_Days")
    plt.xlabel(col)
    plt.ylabel("Poor_Mental_Health_Days")
    plt.tight_layout()
    plt.show()
    print(f"{col}: Pearson correlation = {corr:.3f}")
```

 Looking at the Pearson correlation values, some features show strong correlations ( Insufficient Sleep: 0.689, Food Insecurity: 0.615), while others have weak or negative correlations (e.g., Demographic Vulnerability: -0.054, Mental Health Providers: -0.043).

**Standardize the data**

Stanndardizing the data using the formula (input_feats-mean)/std_deviation make sures all the features are on the same scale. so that no variable dominates the model. It helps the models to perform better.

Standardizing after splitting the data ensures that the test data remain unseen during the training preventaing the data leakage. If Standardization is done before splitting the mean and standard deviaiton from the whole data set ould lead information about test data leading to biased results. This helps to keep model evaluation fair and realistic when predicting the new data.

```python
# -- calulating  mean,standard-deviation
train_means = feat_tr.mean(axis=0)
train_stds  = feat_tr.std(axis=0)

# -- standardizing the data by calulating  (x- mean / standard-deviation)
feat_tr = (feat_tr - train_means) / train_stds
feat_te = (feat_te - train_means) / train_stds
```

**Random Forest Regression**

I chose Random Forest Regression for my county-level mental health dataset because it naturally handles non-linear relationships and mixed data types without forcing a linear dependency between predictors and the outcome. My EDA revealed a mix of strong, moderate, and even negative Pearson correlations across various socio-economic and health-behavior factors, and Random Forests are flexible enough to incorporate all these predictors without having to pre-specify their relationships. They reduce overfitting by averaging results from many bootstrapped decision trees, which smooths out the noise inherent in smaller counties or areas with extreme values.

Additionally, Random Forests offer built-in mechanisms for feature selection by ranking predictors based on their contribution to reducing error. This is particularly valuable when dealing with multiple variables that might be operating on different scales or have been imputed. The algorithm's robustness to outliers and missing data is another major advantage in my dataset, where genuine extremes in values can otherwise distort individual decision trees.

Moreover, because Random Forests require minimal hyperparameter tuning to achieve solid baseline performance, I can quickly iterate on model development. They also provide internal validation measures, such as out-of-bag error estimates, which help ensure that the model's predictions will generalize well to new data. All these benefits make Random Forest the ideal choice for effectively modeling and understanding the complex relationships present in county-level mental health outcomes.

Our improved vs. worsened label is very unbalanced: most counties are “worsened” and only a few are “improved.” If we just split and train, the model can guess “worsened” all the time and still be right most of the time.

To fix this, we can try three things:

1. Drop some “worsened” examples so both groups are the same size.

   * It is easy but we throw away real data.

2. Make new “improved” examples from the ones we have (SMOTE).

   * We keep all the real data and add more “improved” cases.
   * But the new ones might be noisy if the groups overlap.

3. Tell the model to care more about “improved” errors (class weights).

   * We do not remove or invent data.
   * The model learns that missing an “improved” case is worse.

Dropping cases left us too little data, so we should use SMOTE to boost the “improved” group and also give extra weight to “improved” errors. This way the model sees enough “improved” examples and learns not to ignore them.


```python
# -- creating list of input features
feat_cols = [
       'Physical_Inactivity_pct',
       'Adult_Smoking_pct',
       'Adult_Obesity_pct',
       'Access_to_Exercise_Opportunities_pct',
       'Excessive_Drinking_pct',
       'Food_Insecurity_pct',
       'Insufficient_Sleep_pct',
      ]

# -- create input features
feat = diff_2019_2023[feat_cols]
feat.columns

# -- create target features
targ = diff_2019_2023['MH_change_Code']
```

SMOTE (Synthetic Minority Over-sampling Technique)
Generates new minority-class examples by interpolating between existing ones, balancing the dataset without simple duplication.
 Helps the model learn decision boundaries in regions where the minority class is under-represented.

we apply only to the training set
Prevents synthetic data from leaking into validation or test splits, so performance estimates remain unbiased.
Ensures the classifier sees equal class representation during training, improving its ability to detect the minority class.

####**Hyper parameters tuning using the Grid Search CV method**

Tuning tree hyperparameters via GridSearchCV helps to strike a better bias-variance balance than the default settings:

Control model complexity: by testing different max_depth, min_samples_leaf and min_samples_split values, we prevent the tree from growing too deep (overfitting noise) or staying too shallow (underfitting signal).

GridSearchCV runs five-fold CV for each parameter combination, evaluates them on our minority-class recall scorer, and picks the one that best catches “improved” counties without memorizing the training data.

Better generalization when compared to the one-shot default tree, the tuned model usually shows higher recall on unseen data and less disparity between training and test scores, so it reliably identifies more of those rare improved cases in new counties.

```python
recall0 = make_scorer(recall_score, pos_label=0)

# -- select the hyper parameters to tune and the values to try
tuned_parameters = {
   "max_depth": [20, 30, 40, 50],
   "min_samples_leaf": [10, 15, 20, 30],
   "min_samples_split": [2, 5, 10]
}

# -- perform Grid Search
dtc_tune = DecisionTreeClassifier(random_state=302)
cv_tune = GridSearchCV(dtc_tune, tuned_parameters, cv=5, scoring=recall0, verbose=1)
cv_tune.fit(feat_tr,targ_tr)
```
